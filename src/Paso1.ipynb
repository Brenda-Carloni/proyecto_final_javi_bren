## Paso 1: Definición del Problema



### Problema: 

El estrés afecta la vida de las personas en muchos aspectos y a veces se naturalizan sus síntomas. El uso intensivo de plataformas digitales y la hiperconectividad en los estudiantes se suelen asociar con trastornos del sueño, estrés crónico y deterioro de la salud mental. Sin embargo, las instituciones educativas no poseen mecanismos de detección temprana para ayudar a sus alumnos en situaciones vulnerables antes de que colapsen. 

Datos como la exposición digital, los hábitos del sueño y actividad física actúan como indicadores fisiológicos y conductuales tempranos y nos permiten diseñar herramientas no invasivas de evaluación temprana para identificar a personas en situación de vulnerabilidad antes de recurrir a evaluaciones clínicas extensas.

Objetivo:
Diseñar un modelo predictivo capaz de identificar el nivel de estrés en jóvenes o individuos a través de métricas como su tiempo y comportamiento en pantallas, hábitos de descanso y horas de actividad física.
