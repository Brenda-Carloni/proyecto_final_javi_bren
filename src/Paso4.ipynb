Ahora agregaremos la Moda, las medidas de dispersión (Varianza y Desviación Estándar) y las medidas de formas (Curtosis y Asimetría) de cada una de las variables predictoras numéricas para enriquecer nuestro análisis.

```python
variables = [
    'Age', 
    'Avg_Daily_Usage_Hours', 
    'Daily_Unlocks',
    'Study_Hours', 
    'Physical_Activity_Hours', 
    'Sleep_Hours_Per_Night', 
    'Mental_Health_Score'
]

estadisticas = pd.DataFrame({
    "Media": df[variables].mean(),
    "Mediana": df[variables].median(),
    "Moda": df[variables].mode().iloc[0],
    "Desv. estándar": df[variables].std(),
    "Varianza": df[variables].var(),
    "Mínimo": df[variables].min(),
    "Q1": df[variables].quantile(0.25),
    "Q3": df[variables].quantile(0.75),
    "Máximo": df[variables].max(),
    "Asimetría": df[variables].skew(),
    "Curtosis": df[variables].kurt()
})

estadisticas.round(3)
```

- ***Study_Hours:*** El promedio de horas de estudio es de 3,01 horas diarias, aunque existe una dispersión importante (DE = 1,64). El 50% central se encuentra entre 1,5 y 4,2 horas. Presenta una asimetría positiva moderada (0,436), mostrando concentración de valores bajos y una cola hacia valores altos.

En relación a las conclusiones anteriores, nos surge el siguiente contraste de hipótesis:

**1 . Relación entre desbloqueos diarios y nivel de estrés: ¿Hay correlación entre estas variables?**


H₀: No existe una relación significativa entre la cantidad de desbloqueos diarios (Daily_Unlocks) y el nivel de estrés (Stress_Level) de un estudiante.

H₁: Existe una relación significativa entre la cantidad de desbloqueos diarios (Daily_Unlocks) y el nivel de estrés (Stress_Level) de un estudiante.

Dado que Daily_Unlocks es una variable cuantitativa y Stress_Level es una variable categórica, se puede evaluar esta hipótesis comparando la distribución de Daily_Unlocks entre las diferentes categorías de Stress_Level. Dado que Stress_Level tiene más de dos categorías, la prueba adecuada es ANOVA de un factor:

```python
from scipy.stats import f_oneway

grupos = [
    grupo["Daily_Unlocks"].dropna()
    for _, grupo in df.groupby("Stress_Level")
]

resultado = f_oneway(*grupos)

print("Estadístico F:", resultado.statistic)
print("p-value:", resultado.pvalue)
```

**Conclusión**

Podemos observar que nuestro p-value es menor a 0.05, por ende se rechaza la hipótesis nula (H₀) y se acepta la hipótesis alternativa, confirmando que existe evidencia estadísticamente significativa de una relación entre la cantidad de desbloqueos diarios (**Daily_Unlocks**) y el nivel de estrés (**Stress_Level**). Las diferencias observadas en la cantidad de desbloqueos diarios entre los distintos niveles de estrés son estadísticamente significativas.
